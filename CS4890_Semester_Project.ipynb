```python
import re, unicodedata, os

def normalize(s):
  s = unicodedata.normalize("NFKC", s)
  s = s.replace("_eou_", " ")
  s = s.replace("’", "'").replace("‘", "'").replace("“", '"').replace("”", '"')
  # Remove all punctuation and special characters, leaving only alphanumeric characters and whitespace
  s = re.sub(r"[^\w\s]", "", s)
  # Replace consecutive whitespace characters with a single space
  s = re.sub(r"\s+", " ", s).strip()
  return s

def key(s):
  return s.lower().strip()

files = {"american_words.txt", "dialogues_text.txt", "dialogues_train.txt", "lm_test_comm.txt", "lm_test_switch.txt",
         "sent_dev_aac.txt", "sent_test_aac.txt", "sent_train_aac.txt",
         "vocab_aac_twitter.txt"}

for file_path in files:
    if not os.path.exists(file_path):
      print(f"File '{file_path}' not found")
      continue

    with open(file_path, 'r', encoding='utf-8') as file:
      lines = [normalize(l) for l in file]
      lines = [l for l in lines if re.search(r"[A-Za-z]", l) and 1 <= len(l.split()) <=25]

      seen, clean = set(), []
      for l in lines:
        k = key(l)
        if k not in seen:
          seen.add(k); clean.append(l)

    basename = os.path.basename(file_path).replace('.txt', '_normalized_data.txt')
    with open (basename, 'w', encoding='utf-8') as outfile:
      for l in clean:
        outfile.write(l + '\n')
      print(f"Wrote {len(clean)} lines to {basename}")
```
